
Рентабельность активов

```python
pb = 910
pp = 890
of = 2500
cf = 2600

l = 2300
c = 1.5
cc = 1.1

r_a = (pb) / (of + cf) * 100
print(f"Рентабельность активов = {r_a}%")
```


Рентабельность продаж

```python
r_p = pp / (l * c) * 100
print(f"Рентабельность продаж = {r_p}%")
```


Рентабельность продукции

```python
r_p_p = pp / (cc * l) * 100
print(f"Рентабельность продукции = {r_p_p}%")
```
