## 1 этап

### 1.1 Определение стратегической цели – прогнозируемого объема исходящих международных телефонных разговоров (мтр) на конец планируемого периода (10-й год) по формуле 15.1


```python
Y_str = 30.45
N_pl = 139.65
q_str = Y_str * N_pl
q_str
```

### 1.2 Оценка ожидаемых результатов – исходящего обмена на конец планируемого периода (10-й год) методом экстраполяции по формулам 15.2. и 15.3


```python
import math
Q_i = [1271.55, 1362.9, 1409.1, 1475.25, 1572.9]
#q2/q1 * q3/q2 * q4/q3 * q5/q4 = q5/q1
J = (Q_i[4] / Q_i[0]) ** (1/4)
J
```

```python
q_ecstr = Q_i[-1] * J ** 5
print(f"Ожидаемые результаты = {q_ecstr}")
```

### 1.3. Определение стратегического разрыва – прироста исходящего обмена (мтр) для достижения прогнозируемой нормы потребления услуг (формула 15.4.)


```python
delta_q = q_str - q_ecstr
print(f"Стратегический разрыв = {delta_q}")
```

## 2 этап

### 2.1. Прирост обмена за счет повышения эффективности использования мощностей определяется по формуле


```python
M_z = 346.5
C_n = 19.85
C_f = 17.22
dC = C_n - C_f
dq_dc = M_z * dC
print(f"Прирост обмена за счет повышения эффективности использования мощностей {dq_dc}")
```

```python
M_mon = 420
M_z = 346.5
dM = M_mon - M_z
dq_dm = dM * C_f
print(f"Прирост обмена за счет лучшего освоения производственных мощностей = {dq_dm}")
```

```python
dq_int = (dq_dc + dq_dm) * 0.52
print(f"Определение прироста исходящего обмена за счет интенсификации производства = {dq_int}")
```
