```python
q = 10000
s = 80
price = s * 1.25
co = 50000
t = 360

ko = q * price / co
to = t / ko
kz = 1 / ko

print("Коэф. оборачиваемости об. средств: ", ko)
print("Показатель средней продолжительности одного оборота в днях", to)
print("Коэффициент закрепления (загрузки) оборотных средств", kz)
```
